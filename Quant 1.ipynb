```python
def top_of_book(
    events: list[tuple[str, int, str, int, int] | tuple[str, int]],
) -> list[tuple[int | None, int | None]]:
    live_orders = {}
    results = []

    for event in events:
        if event[0] == "add":
            order_id = event[1]
            side = event[2]
            price = event[3]

            live_orders[order_id] = (side, price)

        elif event[0] == "cancel":
            order_id = event[1]

            if order_id in live_orders:
                del live_orders[order_id]

        bid_prices = []
        ask_prices = []

        for side, price in live_orders.values():
            if side == "B":
                bid_prices.append(price)
            elif side == "S":
                ask_prices.append(price)

        best_bid = max(bid_prices) if bid_prices else None
        best_ask = min(ask_prices) if ask_prices else None

        results.append((best_bid, best_ask))

    return results
```
